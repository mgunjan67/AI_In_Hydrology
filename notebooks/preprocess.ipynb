```python
import pandas as pd
final_df = pd.read_csv('final.csv')
static_df = pd.read_csv('static_data.csv')
parameters_df = pd.read_csv('parameters.csv')
```

```python
final_df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
# Step 1: Convert Date to datetime
final_df['Date'] = pd.to_datetime(final_df['Date'])

# Step 2: Handle Missing Values
# Check for missing values
missing_values = final_df.isnull().sum()
print("Missing values per column before handling:\n", missing_values)
```

```python
# Step 3: Create Lag Features
# Create a 1-day lag for runoff value, which can help capture recent runoff patterns
final_df['runoff_lag1'] = final_df['runoff_value'].shift(1)

weather_columns = ['601_ppt', '604_ppt', '605_ppt', '606_ppt', '715_ppt', '805_ppt', '810_ppt', '814_ppt', 
                   '601_tmax', '601_tmin']
# Lag features for precipitation data (adjust as needed based on patterns)
for col in weather_columns:
    final_df[f'{col}_lag1'] = final_df[col].shift(1)

# Drop any rows with NaNs that might have been introduced by lagging
final_df = final_df.dropna()

# Step 4: Add Seasonal Features
final_df['month'] = final_df['Date'].dt.month  # Month can capture seasonality
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Extract year and month for grouping
final_df['Year'] = final_df['Date'].dt.year
final_df['Month'] = final_df['Date'].dt.month

# Month groupings for each subplot
month_groups = [(1, 3), (4, 6), (7, 9), (10, 12)]
titles = ['January - March', 'April - June', 'July - September', 'October - December']

# Generate and save each plot
for i, (start_month, end_month) in enumerate(month_groups):
    plt.figure(figsize=(10, 6))
    
    # Filter data for each set of 3 months
    for month in range(start_month, end_month + 1):
        month_data = final_df[final_df['Month'] == month]
        plt.plot(month_data['Date'], month_data['runoff_value'], label=f'Month {month}', alpha=0.7)
    
    # Set titles and labels
    plt.title(f'Runoff Value Over Time ({titles[i]})')
    plt.xlabel('Date')
    plt.ylabel('Runoff Value')
    plt.legend(title='Month', loc='upper right')
    plt.grid(True)
    
    # Save the figure
    plt.tight_layout()
    plt.savefig(f'runoff_plot_{titles[i].replace(" ", "_")}.png')
    plt.close()  # Close the figure to free memory

print("Plots saved successfully.")
```

```python
# Step 1: Reshape final_df to a long format
# Extract the station numbers and variable types
id_vars = ['Date', 'runoff_value']  # Columns to keep in place
value_vars = [col for col in final_df.columns if col not in id_vars]

# Melt the DataFrame to create 'Station' and 'Type' columns
long_df = pd.melt(final_df, id_vars=id_vars, value_vars=value_vars,
                  var_name='Station_Type', value_name='Value')

# Split 'Station_Type' into 'Station' and 'Type'
long_df[['Station', 'Type']] = long_df['Station_Type'].str.extract(r'(\d+)_([a-zA-Z]+)')

# Drop rows where 'Station' is NaN
long_df = long_df.dropna(subset=['Station'])

# Convert 'Station' to integer after handling NaNs
long_df['Station'] = long_df['Station'].astype(int)

# Step 2: Merge with static data on 'Station'
merged_df = pd.merge(long_df, static_df, left_on='Station', right_on='Index_No_', how='left')
merged_df = merged_df.drop(columns='Index_No_')  # Drop redundant column

# Step 3: Optionally add parameter data
# merged_df = merged_df.merge(parameters_df, left_on='Type', right_on='Parameter Name', how='left')

# Save the merged data if necessary
merged_df.to_csv('merged_final_static.csv', index=False)

print("Data merged successfully. Check 'merged_final_static.csv' for the output.")
```
