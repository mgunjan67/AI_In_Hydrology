```python
import pandas as pd
runoff_data = pd.read_csv('runoff_data.csv')
```

```python
runoff_data.info()
```

```python
runoff_data.head()
```

```python
runoff_data = runoff_data.loc[:, ~runoff_data.columns.str.contains('^Unnamed')]
runoff_data.head()
```

```python
runoff_data.to_csv('runoff_data.csv')
```

```python
time_series_data = pd.read_csv('time_series_data.csv')
time_series_data.info()
```

```python
# Merge the datasets on Date
combined_data = pd.merge(runoff_data[['Date', 'runoff_value']], time_series_data, on='Date', how='inner')
combined_data.info()
```

```python
combined_data.head()
```

```python
combined_data = combined_data.loc[:, ~combined_data.columns.str.contains('^Unnamed')]
combined_data.head()
```

```python
combined_data.to_csv('final_data.csv')
```

```python
combined_data.info()
```

```python
combined_data.replace('#VALUE!', pd.NA, inplace=True)
combined_data.info()
```

```python
import numpy as np
# Replace specific non-numeric values like '#VALUE!' with NaN
combined_data['605_tmax'].replace('#VALUE!', np.nan, inplace=True)

# Convert to numeric with errors coerced to NaN
combined_data['605_tmax'] = pd.to_numeric(combined_data['605_tmax'], errors='coerce')

# Now fill NaN values with the column mean
combined_data['605_tmax'].fillna(combined_data['605_tmax'].mean(), inplace=True)

# Confirm the changes
print(combined_data.info())
```

```python
combined_data.to_csv('final.csv')
```
