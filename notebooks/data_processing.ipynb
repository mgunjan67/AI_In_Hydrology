```python
tmax = pd.read_csv('Tmax.csv')
tmin = pd.read_csv('tmin.csv')
```

```python
print(tmax.info())
print(tmin.info())
```

```python
merged_df = pd.merge(tmax, tmin, on='Date', suffixes=('_tmax', '_tmin'))

# Display the merged dataframe
print(merged_df)
```

```python
merged_df.info()
```

```python
merged_df.head()
```

```python
# List to store valid columns (where both Tmax and Tmin exist)
valid_columns = ['Date']  # Always keep the Date column

# Check for each station if both Tmax and Tmin columns exist
for station in index_no_values:
    tmax_col = f'ST_{station}_tmax'
    tmin_col = f'ST_{station}_tmin'
    
    # Only include the station if both Tmax and Tmin columns exist
    if tmax_col in merged_df.columns and tmin_col in merged_df.columns:
        valid_columns.append(tmax_col)
        valid_columns.append(tmin_col)

# Filter the dataframe to only keep valid columns
filtered_df = merged_df[valid_columns]
```

```python
filtered_df.head()
```

```python
filtered_df.info()
```

```python
t_filtered = filtered_df
t_filtered.info()
```

```python
ppt = pd.read_csv('ppt.csv')
ppt.head()
```

```python
ppt.info()
```

```python
ppt = ppt.iloc[:, :-1]
```

```python
ppt.info()
```

```python
ppt.head()
```

```python
# Create the list of column names you want (e.g., 'st601', 'st604', etc.)
filtered_columns = [f"st{index}" for index in index_no_values]

# Select only the filtered columns (plus the 'Date' column if needed)
df_filtered = ppt[['Date'] + filtered_columns] if 'Date' in ppt.columns else ppt[filtered_columns]
df_filtered.info()
```

```python
df_filtered.head()
```

```python
ppt_filtered = df_filtered
ppt_filtered.info()
```

```python
static_data.head()
```

```python
ppt_filtered.head()
```

```python
ppt_filtered.columns = ppt_filtered.columns.str.replace('st', '')
ppt_filtered.head()
```

```python
t_filtered.columns = t_filtered.columns.str.replace('ST_', '')
t_filtered.head()
```

```python
# Merging the two DataFrames on 'Date'
merged_df = pd.merge(ppt_filtered, t_filtered, on='Date', how='inner')
merged_df.head()
```

```python
merged_df.info()
```

```python
# Rename columns to include '_ppt' suffix for ppt columns
merged_df.rename(columns=lambda x: x + '_ppt' if x in ['601', '604', '605', '606', '715', '805', '810', '814'] else x, inplace=True)

# Create a list for the new column order
new_order = [
    'Date', 
    '601_ppt', '601_tmax', '601_tmin',
    '604_ppt', '604_tmax', '604_tmin',
    '605_ppt', '605_tmax', '605_tmin',
    '606_ppt',
    '715_ppt', '715_tmax', '715_tmin',
    '805_ppt', '805_tmax', '805_tmin',
    '810_ppt', '810_tmax', '810_tmin',
    '814_ppt', '814_tmax', '814_tmin'
]

# Reindex the DataFrame to the new column order
df = merged_df.reindex(columns=new_order)
df.info()
```

```python
time_series_data = df
time_series_data.to_csv('time_series_data.csv')
```
